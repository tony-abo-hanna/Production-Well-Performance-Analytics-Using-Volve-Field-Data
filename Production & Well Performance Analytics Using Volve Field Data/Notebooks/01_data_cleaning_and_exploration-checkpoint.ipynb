# 4. Data Cleaning & Preprocessing

Data cleaning is performed to improve analytical quality and prepare the dataset for engineering analysis and dashboarding.

## 4.1 Standardize Column Names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
)
```

## 4.2 Convert Date Column

```python
df['dateprd'] = pd.to_datetime(df['dateprd'])
```

## 4.3 Evaluate Missing Values

```python
df.isnull().sum().sort_values(ascending=False)
```

### Observations

Missing values are mainly concentrated in:
- choke size
- downhole pressure
- gas/oil operational parameters

These are common in real production datasets due to:
- sensor interruptions
- shut-in periods
- operational downtime

## 4.4 Select Production & Operational Variables

```python
production_df = df[
    [
        'dateprd',
        'npd_well_bore_name',
        'bore_oil_vol',
        'bore_gas_vol',
        'bore_wat_vol',
        'avg_downhole_pressure',
        'avg_choke_size_p',
        'flow_kind'
    ]
]
```

## 4.5 Check Duplicate Records

```python
production_df.duplicated().sum()
```

## 4.6 Handle Missing Production Values

```python
production_df[
    ['bore_oil_vol', 'bore_gas_vol', 'bore_wat_vol']
].isnull().sum()
```

```python
production_df.loc[
    :,
    ['bore_oil_vol', 'bore_gas_vol', 'bore_wat_vol']
] = production_df[
    ['bore_oil_vol', 'bore_gas_vol', 'bore_wat_vol']
].fillna(0)
```

```python
production_df.loc[:, 'avg_downhole_pressure'] = (
    production_df['avg_downhole_pressure']
    .ffill()
)
```

### Observations

Missing production values were replaced with zero where appropriate to preserve production continuity.

Pressure values were forward-filled to maintain operational trends without introducing unrealistic spikes.

## 4.7 Remove Invalid Negative Production Records

```python
production_df = production_df[
    (production_df['bore_oil_vol'] >= 0) &
    (production_df['bore_gas_vol'] >= 0)&
    (production_df['bore_wat_vol'] >= 0)
]
```

# 5. Feature Engineering

Additional engineering metrics were created to support:
- production surveillance
- reservoir maturity analysis
- dashboard KPIs

## 5.1 Create Water Cut Feature

```python
production_df['water_cut'] = (
    production_df['bore_wat_vol'] /
    (
        production_df['bore_oil_vol'] +
        production_df['bore_wat_vol']
    )
)
```

## 5.2 Create Total Liquid Feature

```python
production_df['total_liquid'] = (
    production_df['bore_oil_vol'] +
    production_df['bore_wat_vol']
)
```

## 5.3 Create Gas-Oil Ratio (GOR)

```python
production_df['gor'] = (
    production_df['bore_gas_vol'] /
    production_df['bore_oil_vol']
)
```

## 6.1 Field Oil Production Trend

This analysis evaluates overall field oil production behavior over time to identify:
- field development stages
- production plateau periods
- long-term decline behavior
- operational fluctuations

```python
field_oil = (
    production_df
    .groupby('dateprd')['bore_oil_vol']
    .sum()
)
```

```python
plt.figure(figsize=(14,6))

field_oil.plot()

plt.title('Field Oil Production Over Time')
plt.xlabel('Date')
plt.ylabel('Oil Production Volume')

plt.show()
```

### Observations

- Field production increased rapidly during the early development period.
- Peak production occurred during the initial field ramp-up stage.
- A gradual long-term decline trend is visible after the production plateau.
- Several sharp fluctuations may indicate operational interruptions, maintenance activities, or shut-in periods.
- The overall trend reflects a mature producing field undergoing depletion.

## 6.2 Total Oil Production by Well

This analysis compares cumulative oil contribution from individual wells to identify:
- top-performing wells
- production inequality across wells
- dominant field contributors

```python
well_oil = (
    production_df
    .groupby('npd_well_bore_name')['bore_oil_vol']
    .sum()
    .sort_values(ascending=False)
)
```

```python
plt.figure(figsize=(14,6))

well_oil.plot(kind='bar')

plt.title('Total Oil Production by Well')
plt.xlabel('Well Name')
plt.ylabel('Total Oil Production')

plt.xticks(rotation=45)

plt.show()
```

### Observations

- A small number of wells contribute the majority of total field production.
- Production contribution varies significantly between wells.
- Several wells show relatively low cumulative production, likely due to shorter operational life or poorer reservoir performance.

## 6.3 Cumulative Oil Production by Well

Cumulative production trends are analyzed to evaluate long-term well productivity and reservoir drainage behavior.

```python
selected_wells = well_oil.index[:5]
```

```python
plt.figure(figsize=(14,7))

for well in selected_wells:

    well_data = production_df[
        production_df['npd_well_bore_name'] == well
    ]

    cumulative_oil = (
        well_data
        .sort_values('dateprd')['bore_oil_vol']
        .cumsum()
    )

    plt.plot(
        well_data.sort_values('dateprd')['dateprd'],
        cumulative_oil,
        label=well
    )

plt.title('Cumulative Oil Production by Well')
plt.xlabel('Date')
plt.ylabel('Cumulative Oil Production')

plt.legend()

plt.show()
```

### Observations

- High-performing wells exhibit steady cumulative production growth over time.
- Differences in cumulative production indicate variability in reservoir quality and well performance.
- Some wells show early flattening trends, suggesting production decline or operational limitations.
- Long-term cumulative behavior helps identify dominant producers within the field.

## 6.19 Field Oil Production Decline Trend

```python
monthly_oil = (
    production_df
    .groupby('dateprd')['bore_oil_vol']
    .sum()
)
```

```python
rolling_oil = monthly_oil.rolling(window=6).mean()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    monthly_oil.index,
    monthly_oil,
    alpha=0.4,
    label='Monthly Production'
)

plt.plot(
    rolling_oil.index,
    rolling_oil,
    linewidth=3,
    label='6-Month Rolling Average'
)

plt.title('Field Oil Production Decline Trend')
plt.xlabel('Date')
plt.ylabel('Oil Production')

plt.legend()

plt.show()
```

### Observations

- The rolling average highlights long-term production decline behavior more clearly.
- Field production demonstrates a mature depletion trend after peak production.
- Short-term operational noise is reduced using the rolling average method.
- The decline profile is consistent with reservoir depletion and increasing water production.
